## Cell 1 — Import libraries and upload the dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files

uploaded = files.upload()  # Select netflix_titles.csv
```

```python
# Load and inspect the CSV
df = pd.read_csv("netflix_titles.csv")

print("Rows and columns:", df.shape)
display(df.head())
df.info()

print("\nMissing values before cleaning:")
print(df.isnull().sum())
print("\nDuplicate rows:", df.duplicated().sum())
```

## Cell 2 — Clean and prepare the data

```python
df = df.drop_duplicates().copy()

# Parse dates; invalid dates become missing rather than being guessed.
df["date_added"] = pd.to_datetime(df["date_added"], errors="coerce")

text_cols = [
    "director", "cast", "country", "rating",
    "duration", "listed_in", "description"
]
for col in text_cols:
    df[col] = df[col].fillna("Unknown")

df["year_added"] = df["date_added"].dt.year
df["duration_number"] = pd.to_numeric(
    df["duration"].str.extract(r"(\d+)")[0], errors="coerce"
)

print("Shape after cleaning:", df.shape)
print("Duplicates remaining:", df.duplicated().sum())
print("\nMissing values after cleaning:")
print(df.isnull().sum())
```

## Cell 3 — Answer five data questions

```python
# Q1: How many Movies and TV Shows?
print("Q1 — Movies vs TV Shows")
print(df["type"].value_counts())

# Q2: Which countries have the most titles?
print("\nQ2 — Top countries")
countries = df[df["country"] != "Unknown"]["country"].str.split(", ").explode()
print(countries.value_counts().head(10))

# Q3: Which content ratings are most common?
print("\nQ3 — Top content ratings")
print(df["rating"].value_counts().head(10))

# Q4: Which genres/categories appear most often?
print("\nQ4 — Top categories")
genres = df["listed_in"].str.split(", ").explode()
print(genres.value_counts().head(10))

# Q5: How many titles were added each year?
print("\nQ5 — Titles added by year")
print(df["year_added"].value_counts().sort_index())
```

## Cell 4 — Create six visualizations

```python
# 1. Bar chart: Movies vs TV Shows
df["type"].value_counts().plot(kind="bar", title="Movies vs TV Shows")
plt.xlabel("Content type")
plt.ylabel("Number of titles")
plt.tight_layout()
plt.show()
```

```python
# 2. Line chart: titles added by year
added = df["year_added"].dropna().astype(int).value_counts().sort_index()
added.plot(kind="line", marker="o", title="Titles Added to Netflix by Year")
plt.xlabel("Year added")
plt.ylabel("Number of titles")
plt.tight_layout()
plt.show()
```

```python
# 3. Histogram: release-year distribution
df["release_year"].dropna().plot(
    kind="hist", bins=20, title="Distribution of Release Years"
)
plt.xlabel("Release year")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
# 4. Scatter plot: release year vs year added
tmp = df.dropna(subset=["release_year", "year_added"])
plt.figure(figsize=(7, 4))
plt.scatter(tmp["release_year"], tmp["year_added"], alpha=0.35)
plt.title("Release Year vs Year Added")
plt.xlabel("Release year")
plt.ylabel("Year added")
plt.tight_layout()
plt.show()
```

```python
# 5. Pie chart: Movies vs TV Shows share
df["type"].value_counts().plot(
    kind="pie", autopct="%1.1f%%", title="Share of Movies and TV Shows"
)
plt.ylabel("")
plt.tight_layout()
plt.show()
```

```python
# 6. Heatmap: numeric correlations
cols = ["release_year", "year_added", "duration_number"]
plt.figure(figsize=(6, 4))
sns.heatmap(
    df[cols].corr(numeric_only=True),
    annot=True, cmap="coolwarm", vmin=-1, vmax=1
)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```
